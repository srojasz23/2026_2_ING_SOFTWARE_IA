```python
# Rutina de notas 
nota = float(input("Ingresa tu nota (0-20 :) "))

# utilizar un if para validar nota ingesara en el intervalo[0,20]
if (nota < 0 or nota > 20 ):
    print("La nota ingresada es invalida ")
#utilizar el correspondiente else para asignar la conversion adecuada 
else: 
    if nota >=18:
        nota_letra = "A" 
    elif nota >=16:
        nota_letra = "B"
    elif nota >=14:
        nota_letra = "C"
    elif nota >=12:
        nota_letra = "D"
    else:
        nota_letra = "E"

    print("Calificacion : %s  " %(nota_letra))
```
